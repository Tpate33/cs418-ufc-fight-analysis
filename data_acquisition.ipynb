```python
import polars as pl
ufc_fights = pl.read_csv("data/ufc_fights.csv")
print(ufc_fights.head())
print(ufc_fights.shape)
print(ufc_fights.schema)
```

```python
ufc_stats = pl.read_csv("data/ufcstats_data.csv")
print(ufc_stats.head())
print(ufc_stats.shape)
print(ufc_stats.schema)

for c in ["height", "weight", "reach", "stance", "dob"]:
    uniques = ufc_stats[c].unique().to_list()
    print(f"\n{c} — {len(uniques)} unique values:")
    print([u for u in uniques if not any(ch.isdigit() for ch in str(u))][:15])
```
